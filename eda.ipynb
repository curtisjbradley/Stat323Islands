```python
import pandas as pd

def MAP(bp : str ):
    sbp, dbp = bp.split('/')
    return (int(sbp) + 2 * int(dbp))/3

results = pd.read_csv('blood_pressure_experiment/blood_pressure_results.csv')

bp_columns = [r for r in  results.columns if 'blood_pressure' in r]

for col in bp_columns:
    results[col] = results[col].map(MAP)

results.to_csv('map_results_with_subsampling.csv', index=False)

results['average_base_map']= (results['base_blood_pressure_1'] + results['base_blood_pressure_2']) / 2

results['average_end_map']= (results['end_blood_pressure_1'] + results['end_blood_pressure_2']) / 2

results.drop(bp_columns, axis=1, inplace=True)

results.to_csv('average_maps.csv', index=False)

results['average_delta_map'] = results['average_end_map'] - results['average_base_map']


results.drop(['average_base_map', 'average_end_map'], axis=1, inplace=True)
results.to_csv('average_delta_map_results.csv', index=False)
```
